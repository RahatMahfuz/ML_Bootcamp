```python
#Importing analysis libraries

import pandas as pd
import numpy as np

import matplotlib as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder,StandardScaler
from sklearn.impute import SimpleImputer

from sklearn.metrics import accuracy_score,classification_report
from sklearn.metrics import mean_squared_error,r2_score
```

```python
#Loading dataset
dataset=pd.read_csv(r"C:\Users\Rahat Mahfuz\Data_Science\ML_Bootcamp\Bootcamp_Assignment\Assignment_2\synthetic_noisy_dataset_40k_15col.csv")
```

```python
#Showing the dataset
dataset
```

```python
#dataset info
dataset.describe()
```

```python
#Data Cleaning

#missing values
dataset.isnull().sum()
```

```python
#numerical columns
numerical_columns=dataset.select_dtypes(include=np.number).columns

imputer=SimpleImputer(strategy='median')
dataset[numerical_columns]=imputer.fit_transform(dataset[numerical_columns])
```

```python
#categorical columns
categorical_columns=dataset.select_dtypes(include='object').columns

imputer_category=SimpleImputer(strategy='most_frequent')
dataset[categorical_columns]=imputer_category.fit_transform(dataset[categorical_columns])
```

```python
#removing duplicates
dataset.drop_duplicates(inplace=True)
dataset
```

```python
#handling invalid categories
for columns in categorical_columns:
    dataset[columns].str.strip().str.lower()
```

```python
#Outlier detection
for columns in numerical_columns:
    Q1=dataset[columns].quantile(1/4)
    Q3=dataset[columns].quantile(3/4)
    IQR=Q3-Q1
    
    lower_limit=Q1-1.5*IQR
    upper_limit=Q3+1.5*IQR
    
    dataset=dataset[(dataset[columns]>=lower_limit)&(dataset[columns]<=upper_limit)]
    
    dataset
```

```python
#identifying problem type
dataset.columns
target_salary='target_salary'
dataset[target_salary].dtype

#Here target salary is a numerical value.So it will be an Regression Model.
```

```python
#Encoding categorical values
encoding=LabelEncoder()

for columns in categorical_columns:
    dataset[columns]=encoding.fit_transform(dataset[columns])
```

```python
#splitting features
x=dataset.drop(target_salary,axis=1)
y=dataset[target_salary]
```

```python
#train-test and split
x_trained,x_test,y_train,y_test=train_test_split(
    x,y,
    test_size=0.2,
    random_state=42
)
```

```python
#feature scaling
number_columns=x_trained.select_dtypes(include="number").columns

scaling=StandardScaler()

x_trained[number_columns]=scaling.fit_transform(x_trained[number_columns])
x_test[number_columns]=scaling.transform(x_test[number_columns])
```

```python
#model selection
#the type of data is float64.So,it is a numerical value.It will be a regression model
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
```

```python
#training the model
linear_regression=LinearRegression()
decision_tree=DecisionTreeRegressor()
random_forest=RandomForestRegressor()

linear_regression.fit(x_trained,y_train)
decision_tree.fit(x_trained,y_train)
random_forest.fit(x_trained,y_train)
```

```python

#prediction
prediction_1=linear_regression.predict(x_test)
prediction_2=decision_tree.predict(x_test)
prediction_3=random_forest.predict(x_test)
```

```python
#evaluation
#choosing the best model with the highest percentage of accuracy
print("Linear Regression")
print(r2_score(y_test,prediction_1))

print("Decision Tree")
print(r2_score(y_test,prediction_2))

print("Random Forest")
print(r2_score(y_test,prediction_3))
```

```python
#The highest score is fro the Random Forest Model.Which is 86.69% of accuracy.So,between these 3 models Random Forest is the best model.
```
